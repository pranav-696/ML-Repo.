```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import gradio as gr
```

```python
df = pd.read_csv("heart_disease_data.csv")
```

```python
# Checking for missing values
df.fillna(df.mean(), inplace=True)
```

```python
# Defineing features and target variable
X = df.drop(columns=['target'])  # Features
Y = df['target']  # Labels
```

```python
# Spliting dataset
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=42)
```

```python
# Standardizing data (Important for Deep Learning models)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

Building the Model

```python
# Building the Deep Learning Model
model = Sequential([
    Dense(16, activation='relu', input_shape=(X_train.shape[1],)),
    Dropout(0.3),  # Prevent Overfitting
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')  # Sigmoid for binary classification
])
```

```python
# Compile Model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the Model
model.fit(X_train, Y_train, epochs=100, batch_size=16, validation_data=(X_test, Y_test), verbose=1)
```

```python
# Evaluateing Model
train_acc = model.evaluate(X_train, Y_train, verbose=0)[1]
test_acc = model.evaluate(X_test, Y_test, verbose=0)[1]
```

```python
print(f"✅ Training Accuracy: {train_acc*100:.2f}%")
print(f"✅ Testing Accuracy: {test_acc*100:.2f}%")
```

```python
# Step 5: Function to Predict Heart Disease
def predict_heart_disease(*features):
    # Convert input into a NumPy array
    input_data = np.array(features).reshape(1, -1)
    input_data = scaler.transform(input_data)  # Apply same scaling

    # Get prediction
    prediction = model.predict(input_data)[0][0]

    # Return result
    return " High Risk - Visit a Doctor" if prediction > 0.5 else " No Heart Disease"
```

```python
# Create a Gradio Web Interface
inputs = [gr.Number(label=f"Feature {i+1}") for i in range(X.shape[1])]
interface = gr.Interface(fn=predict_heart_disease, inputs=inputs, outputs="text", title="Heart Disease Prediction")
```

```python
# Launching the Interface
interface.launch(share=True)  # Generates a public URL for testing
```
